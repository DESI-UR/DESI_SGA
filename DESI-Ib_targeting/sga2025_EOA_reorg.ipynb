We selected the EOA targets based off an early version of the SGA-2025.  Here, we go through the selected EOA targets, update their ID number (`ref_id` to `SGAID`) and identify which galaxies still need targets.

```python
from astropy.io import fits
from astropy.table import Table, vstack
from astropy import units as u
from astropy.coordinates import SkyCoord

import json
```

# SGA-2025 large galaxies

Those with a minor axis greater than the patrol radius of a DESI fiber.

```python
large_galaxies = Table.read('SGA2025_large_galaxies.fits')

large_galaxies[:5]
```

## Add column to track if galaxy's EOA targets were already selected

```python
large_galaxies['EOA'] = False
```

# Old files

Let's check to see if `ref_id` is `SGAID`

```python
old_file = 'SGA2025_large_galaxies_0-old.fits'

old_gals = Table.read(old_file)

old_gals[:5]
```

```python
lg_gal_idx = large_galaxies['SGAID'] == old_gals['ref_id'][4]

large_galaxies['RA', 'DEC', 'PA', 'BA', 'D26'][lg_gal_idx]
```

Cool!  It looks like `ref_id` == `SGAID`, so I'll just mush together all the old EOA target files and reparse based on the new targeting files.

# Smash together "old" EOA target files

```python
EOA_target_dir = 'target_files/'

EOA_target_filenames = 'SGA2025_off-axis_targets_{}.txt'

all_targets = {}

for i in range(10):

    if i != 8:
        infile = open(EOA_target_dir + EOA_target_filenames.format(i), 'r')
        curr_target_list = json.load(infile)
        infile.close()
    
        all_targets.update(curr_target_list)
```

```python
outfile = open(EOA_target_dir + 'SGA2025_off-axis_targets_old.txt', 'w')

json.dump(all_targets, outfile)

outfile.close()
```

# Smash together new lists of galaxies that need targeting

```python
EOA_filenames = 'SGA2025_large_galaxies_{}.fits'

all_tables = [Table.read(EOA_filenames.format(i), format='fits') for i in range(11)]

all_gals = vstack(all_tables)
```
