```python
path = '/pscratch/sd/d/dbustos/loa_cutouts/cutouts_fibers'
ids = np.unique(quick_VI['SGA_ID'])
batch_size = 20
```

```python
for b in range(0,len(ids),batch_size)[:0]:
    batch_ids = ids[b:b + batch_size]

    fig, ax = plt.subplots(10, 2, figsize = (16,40))

    ax = ax.flatten()

    for idx, i in enumerate(batch_ids):
        img_file = os.path.join(path, 'cutouts_sga_'+ str(i)+'.png')

        if os.path.exists(img_file):
            img = mpl.image.imread(img_file)
            ax[idx].imshow(img)
        else:
            ax[idx].text(.5,.5, 'Missing')

        ax[idx].axis('off')

    plt.tight_layout()

    plt.show()
```
