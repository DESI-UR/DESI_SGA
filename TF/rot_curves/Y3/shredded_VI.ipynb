```python
counter = 0

for file in os.listdir(path)[:0]:

    counter += 1

    print(counter)
    
    img_file = os.path.join(path, file)
    
    img = mpl.image.imread(img_file)

    file_name = os.path.splitext(file)
    
    sga_idx = int(file_name[0].split('_')[2])

    # missing_targets_path = '/pscratch/sd/d/dbustos/loa_cutouts/cutouts_fibers_missing/cutouts_sga_{}'.format(sga_idx)

    # if os.path.exists(missing_targets_path):
    #     print (sga_idx, ': in missing fibers folder')
    #     continue

    targets = shredded[shredded['SGA_ID'] == sga_idx]['TARGETID']
    if len(targets) >= 10:
        print(sga_idx, ': too many targets')
        flagged_files += [img_file]
        continue
    
    fig, ax = plt.subplots(figsize = (12,8))

    ax.clear()
    ax.imshow(img)

    fig.canvas.draw()

    plt.show()

    print('SGA ID:', sga_idx)

    for i, j in enumerate(targets):
        print(i, ': ',j)

    user_ipt = input('f to flag, g to remove galaxy, or enter target id numbers, or s to continue: ').strip()
    
    if user_ipt == 'f':
        flagged_files += [img_file]
        print('flagged', file)

    elif user_ipt == 'g':
        sga_id += [sga_idx]
        print('added {} to remove list'.format(sga_idx))
    
    elif user_ipt == 's':
        print('next galaxy')
        continue

    elif user_ipt:

        remove = user_ipt.split()
        
        target_ids += [targets[int(r)] for r in remove]

        matched +=  ['{}: {}'.format(sga_idx, targets[int(r)]) for r in remove]
            
        print('added target ids to remove list')

print('finished!')
```

```python
counter = 0

for file in flagged_missing[:0]:

    counter += 1

    print(counter)
    
    img_file = file
    
    img = mpl.image.imread(img_file)

    file_name = os.path.splitext(file)
    
    sga_idx = int(file_name[0].split('_')[5])

    no_fiber_file = os.path.join(no_fiber_path, 'sga_{}.jpg'.format(sga_idx))
    img_no_fiber = mpl.image.imread(no_fiber_file)

    # missing_targets_path = '/pscratch/sd/d/dbustos/loa_cutouts/cutouts_fibers_missing/cutouts_sga_{}'.format(sga_idx)

    # if os.path.exists(missing_targets_path):
    #     print (sga_idx, ': in missing fibers folder')
    #     continue

    targets = shredded_missing[shredded_missing['SGA_ID'] == sga_idx]['TARGETID']
    
    if len(targets) >= 10:
        print(sga_idx, ': too many targets')
        redo_cutout += [sga_idx]
        continue
    
    fig, (ax1, ax) = plt.subplots(2, 1, figsize = (12,16))

    ax.clear()
    ax.imshow(img)

    ax1.clear()
    ax1.imshow(img_no_fiber)

    # fig.canvas.draw()

    plt.show()

    print('SGA ID:', sga_idx)

    for i, j in enumerate(targets):
        print(i, ': ',j)

    user_ipt = input('r: redo cutout, g: remove galaxy, m: manual VI, target id number idx, or s to continue: ').strip()
    
    if user_ipt == 'r':
        redo_cutout += [sga_idx]
        print('will remake cutout')

    elif user_ipt == 'g':
        sga_id += [sga_idx]
        print('added {} to remove list'.format(sga_idx))
    
    elif user_ipt == 's':
        print('next galaxy')
        continue

    elif user_ipt == 'm':
        print('will manually remove points')
        continue

    # elif user_ipt == 'w':
    #     print('will add to weird ellipse')
    #     weird_ellipse += [sga_idx]
    #     sga_id += [sga_idx]
    #     continue
        

    elif user_ipt:

        remove = user_ipt.split()
        
        target_ids += [targets[int(r)] for r in remove]

        matched +=  ['{}: {}'.format(sga_idx, targets[int(r)]) for r in remove]
            
        print('added target ids to remove list')

print('finished!')
```
