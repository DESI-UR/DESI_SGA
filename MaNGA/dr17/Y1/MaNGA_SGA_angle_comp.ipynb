```python
import numpy as np
import matplotlib.pyplot as plt
import numpy.ma as ma

from astropy.table import Table
```

```python
manga_sga = Table.read('/pscratch/sd/s/sgmoore1/TF/sga_manga_rotcurve_comparisons.fits')
```

```python
manga_sga[:5]
```

## get $\Delta \phi$ using the following formula (s.t. $\Delta \phi < 90$)

1. $\phi_{\rm MaNGA} = \phi_{\rm MaNGA}\, \%\, 180$
2. $|\Delta \phi| = min(||\phi_{\rm MaNGA} - \phi_{SGA}| - 180|, |\phi_{\rm MaNGA} - \phi_{\rm SGA}|)$

```python
phi_manga = manga_sga['phi'] % 180
dphi = np.min([np.abs(np.abs(phi_manga - manga_sga['PA']) -180), np.abs(phi_manga - manga_sga['PA'])], axis=0)
```

## $\Delta \phi$ (MaNGA-SGA) vs inclination

```python
plt.hist(dphi, bins=np.linspace(0,90,100))
plt.xlabel(r'$|\Delta \phi|$')
```

```python
plt.scatter(dphi, manga_sga['BA'], alpha=0.1, marker='.')
plt.xlabel(r'$|\Delta \phi|$')
plt.ylabel(r'SGA axis ratio')
```

```python
# edges = np.quantile(dphi, np.linspace(0,1, 41))
# adding medians

bins = np.linspace(0, 90,16)
bins = np.quantile(dphi, np.linspace(0,1,20))
ctrs = []
medians = []
# data = []

indices = np.digitize(dphi, bins)

for i in range(len(bins)):

    medians.append(ma.median(manga_sga['BA'][indices==i]))
    ctrs.append(ma.mean(dphi[indices==i]))
```

```python
plt.scatter(dphi, manga_sga['BA'], alpha=0.1, marker='.')
plt.xlabel(r'$|\Delta \phi|$')
plt.ylabel(r'SGA axis ratio')
plt.scatter(ctrs, medians, marker='x', color='navy')
```
